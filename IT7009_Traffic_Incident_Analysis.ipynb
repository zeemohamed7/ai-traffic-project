### 1.2.1 Missing Value Analysis
This type of analysis looks at how missing values are spread out throughout the records of the data set. The results will help determine an appropriate strategy for handling missing values during *preprocessing*.

```python

# Calculate missing values for each record
missing_counts = df.isna().sum(axis=1)

# Summarize the distribution of missing values
summary = pd.DataFrame({
    "Missing Values": ["None", "1", "2", "3", "Entire Row"],
    "Number of Records": [
        (missing_counts == 0).sum(),
        (missing_counts == 1).sum(),
        (missing_counts == 2).sum(),
        (missing_counts == 3).sum(),
        (missing_counts == df.shape[1]).sum()
    ]
})

# Calculate percentages
summary["Percentage (%)"] = (
    summary["Number of Records"] / len(df) * 100
).round(3)

display(summary)
```

### Justification for removing five isolated anomalies

Five further records were removed as likely data errors following contextual inspection, rather than solely because they exceeded statistical outlier thresholds. The screening rules identified temperatures of −80 and 120, average speed of 350 km/h, lane count of 20 and road width of 120 m. Each value was isolated from the other observations. The thresholds were used to select records for review, not as universal physical limits.

The 350 km/h value represented average traffic speed on a highway with a 100 km/h speed limit, whereas the next highest recorded average was 120 km/h. This combination made the measurement implausible for the traffic conditions represented. The 20-lane observation described an urban road with a recorded width of 4.6 m; all other lane counts were between 1 and 6. The isolated count and its relationship to the recorded width supported treating it as unreliable. A separate six-lane tunnel record had a road width of 120 m, compared with a maximum of 15 m among the other records. Its scale and recorded context supported the same decision.

The temperature values of −80 and 120 were evaluated under the working assumption that Temperature represents ambient air temperature in degrees Celsius. On that basis, their traffic-event context and separation from the remaining observed range of −7.9 to 54.5 supported treating them as likely errors. The Celsius interpretation has not yet been independently confirmed from dataset documentation; this assumption must be verified before final submission, and the two temperature exclusions revisited if it is incorrect.

These removals were documented judgement calls based on the combination of isolation and contextual implausibility, not proof that every flagged value is universally impossible. The correct measurements could not be recovered from the available information, so the five records were removed rather than replaced with guessed values. This reduced the dataset from 49,952 to 49,947 records. Subsequent IQR outliers were retained where there was insufficient additional evidence of erroneous data.

##### Outliers

Step 1 — Detect statistical outliers

```python
# Select continuous numerical measurements
continuous_columns = [
    "Temperature", "Humidity", "Pressure", "Visibility",
    "Wind_Speed", "Precipitation", "Dew_Point",
    "Traffic_Density", "Average_Speed",
    "Heavy_Vehicle_Percentage", "Congestion_Index",
    "Emergency_Response_Time", "Road_Width",
    "Travel_Time_Minutes", "Risk_Score"
]

# Calculate IQR outlier counts for each measurement
for col in continuous_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

```python
# Investigate the four features containing IQR outliers
outlier_columns = [
    "Temperature", "Precipitation",
    "Dew_Point", "Risk_Score"
]

for col in outlier_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Separate lower and upper outliers
    low_values = df_clean.loc[df_clean[col] < lower, col]
    high_values = df_clean.loc[df_clean[col] > upper, col]

    print(f"\n{col}")
    print(f"IQR boundaries: {lower:.2f} to {upper:.2f}")
    print(f"Lower outliers: {len(low_values)}")
    print(f"Upper outliers: {len(high_values)}")
    print(f"Actual range: {df_clean[col].min()} to {df_clean[col].max()}")
```

```python
# Check whether dew point exceeds air temperature - dew point generally cannot exceed air temperature
dew_point_invalid = df_clean["Dew_Point"] > df_clean["Temperature"]

print("Dew point above temperature:", dew_point_invalid.sum())

# Inspect examples of inconsistent measurements
display(
    df_clean.loc[
        dew_point_invalid,
        ["Incident_ID", "Temperature", "Dew_Point", "Humidity"]
    ].head(15)
)
```
