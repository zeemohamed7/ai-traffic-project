### 1.2.1 Missing Value Analysis
This type of analysis looks at how missing values are spread out throughout the records of the data set. The results will help determine an appropriate strategy for handling missing values during *preprocessing*.

```python

# Calculate missing values for each record
missing_counts = df.isna().sum(axis=1)

# Summarize the distribution of missing values
summary = pd.DataFrame({
    "Missing Values": ["None", "1", "2", "3", "Entire Row"],
    "Number of Records": [
        (missing_counts == 0).sum(),
        (missing_counts == 1).sum(),
        (missing_counts == 2).sum(),
        (missing_counts == 3).sum(),
        (missing_counts == df.shape[1]).sum()
    ]
})

# Calculate percentages
summary["Percentage (%)"] = (
    summary["Number of Records"] / len(df) * 100
).round(3)

display(summary)
```

**Observation:** Approximately 99.94% of the dataset records are complete. Five records are entirely empty, while 26 contain between one and three missing values. The completely empty records will be removed during preprocessing, and the remaining incomplete records will be investigated before selecting a treatment method.

##### Outliers

Step 1 — Detect statistical outliers

```python
# Select continuous numerical measurements
continuous_columns = [
    "Temperature", "Humidity", "Pressure", "Visibility",
    "Wind_Speed", "Precipitation", "Dew_Point",
    "Traffic_Density", "Average_Speed",
    "Heavy_Vehicle_Percentage", "Congestion_Index",
    "Emergency_Response_Time", "Road_Width",
    "Travel_Time_Minutes", "Risk_Score"
]

# Calculate IQR outlier counts for each measurement
for col in continuous_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

```python
outlier_columns = [
    "Temperature", "Precipitation", "Dew_Point", "Risk_Score"
]

for col in outlier_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    print("\n" + col)
    print("IQR boundaries:", round(lower, 2), "to", round(upper, 2))
    print("Actual range:", df_clean[col].min(), "to", df_clean[col].max())
    print("Below boundary:", (df_clean[col] < lower).sum())
    print("Above boundary:", (df_clean[col] > upper).sum())
```

This section analyses the cleaned data for identifiers and target revealing columns, encodes categorical variables, and quantifies and plans for the Severity class imbalance.

## 2.1 Leakage Checks, Encoding and Imbalance

###  2.1.1 Loading the cleaned data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

dataset_url = "https://raw.githubusercontent.com/zeemohamed7/ai-traffic-project/main/traffic_incidents_cleaned.csv"


df_fs = pd.read_csv(
    dataset_url,
    parse_dates=["Start_Time", "End_Time"],
    dtype={"Postal_Code": str}
)
df_fs["Severity"] = df_fs["Severity"].astype(int)
print(df_fs.shape)
```

### 2.1.2 Severity class distribution

```python
print(df_fs["Severity"].value_counts().sort_index())
print(df_fs["Severity"].value_counts(normalize=True).sort_index().round(3))

sns.countplot(x="Severity", data=df_fs)
plt.title("Severity Class Distribution")
plt.show()
```

### 2.1.3 Severity imbalance handling plan

Severity is imbalanced (Severity 4 = 55.2%, Severity 1= 1.6%), so accuracy alone is misleading, because a model that always predicts Severity 4 would score about 55%. Plan: 1. Use a stratified split, 2. handle imbalance on the training set only( class weights or SMOTE), never on the test set, and 3. evaluate with macro F1, precision, recall and confusion matrices

### 2.1.4 Leakage check

Findings: Risk_Score (37.8 to 147.3) and Emergency_Response_Time (8.3 to 27.6) rise steadily with Severity, so they are treated as target revealing and are removed. Travel_Time_Minutes, Travel_Category and Response_Team are only known after an incident and are also removed. Accident_History rises with severity (2.6 to 11.7). It would be known before an accident, so it has been kept, but flagged as possible leakage. Congestion_Index stays flat across Severity levels.

```python
check= ["Risk_Score", "Travel_Time_Minutes", "Emergency_Response_Time", "Congestion_Index", "Accident_History"]
print(df_fs.groupby("Severity")[check].mean().round(2))
```

### 2.1.5 drop columns, separate target and encode categories

```python
drop_cols = [
    "Incident_ID","Road_Name", "Postal_Code",
    "Start_Time", "End_Time", "Day", "Year","Weekday",
    "Travel_Time_Minutes", "Travel_Category",
    "Emergency_Response_Time", "Response_Team",
    "Risk_Score", "City", "District"
]

X = df_fs.drop(columns=drop_cols + ["Severity"])
y= df_fs["Severity"]
print(X.shape)
```

### 2.1.6 Encode categories as numbers

```python
X= pd.get_dummies(X, drop_first=True, dtype=int)
print(X.shape)
```

### 2.1.7 Training and Test Split

We use 80% of the data for training and 20% for testing. Stratification keeps the severity proportions similar in both sets. Setting random_state=42 makes the split repeatable.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
print(X.shape)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
print("Training:")
print(y_train.value_counts(normalize=True).sort_index())

print("\nTesting:")
print(y_test.value_counts(normalize=True).sort_index())
```

The training and test sets have similar severity proportions: about 1.6% for Severity 1, 7.0% for Severity 2, 36.3% for Severity 3 and 55.2% for Severity 4. This confirms that the stratified split preserved the class distribution.

#### 2.2.1 Screen for redundency

```python
# Check numeric predictors for strong pairwise correlations.
numeric_features = X_dev.columns[~np.asarray(discrete_features)]
correlations = X_dev[numeric_features].corr().abs()

retained_features = []
redundant_pairs = []

for feature in importance.index:
    if feature in correlations.columns:
        retained_numeric = correlations.columns.intersection(retained_features)
        strong_matches = correlations.loc[feature, retained_numeric].dropna()
        strong_matches = strong_matches[strong_matches >= 0.90]

        if not strong_matches.empty:
            partner = strong_matches.idxmax()
            redundant_pairs.append({
                "Removed feature": feature,
                "Retained feature": partner,
                "Absolute correlation": strong_matches[partner],
            })
            continue

    retained_features.append(feature)

selected_features = retained_features[:20]

display(pd.DataFrame(
    redundant_pairs,
    columns=["Removed feature", "Retained feature", "Absolute correlation"],
).round(3))

display(importance.loc[selected_features].round(4).to_frame())
```

# Unsupervised Learning
In this section, K-Means is implemented to group incidents by traffic conditions without using severity, and to test different numbers of clusters with the elbow method and silhouette score.

```python
condition_features = ["Average_Speed", "Congestion_Index", "Temperature", "Visibility", "Road_Width"]
location_features = ["Latitude", "Longitude"]
clustering_data = df_fs[condition_features + location_features].copy()

clustering_feature_sets = {
    "Traffic conditions": condition_features,
    "Traffic conditions and location": condition_features + location_features,
}
```

## 4.1 Prepare the Clustering Features
We compare traffic condition features, the same features with location, and traffic load with Road_Work. Severity is excluded while clusters are formed. A shuffled copy, made by independently mixing each feature column, provides a check on whether relationships in the real data support clearer groupings.

```python
clustering_data["Traffic_Density"] = df_fs["Traffic_Density"].to_numpy()
clustering_data["Vehicle_Count"] = df_fs["Vehicle_Count"].to_numpy()
clustering_data["Road_Work"] = df_fs["Road_Work"].to_numpy()

clustering_feature_sets["Traffic load and road work"] = ["Traffic_Density", "Vehicle_Count", "Road_Work"]

rng = np.random.default_rng(42)
shuffled_data = clustering_data.copy()
for col in clustering_data.columns:
    shuffled_data[col] = rng.permutation(clustering_data[col].to_numpy())

sample_rows = np.random.default_rng(42).choice(len(clustering_data), size=5000, replace=False)
set_results = []

for name, columns in clustering_feature_sets.items():
    scores = []
    for source in [clustering_data, shuffled_data]:
        model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("kmeans", KMeans(n_clusters=3, n_init=10, random_state=42)),
        ])
        labels = model.fit_predict(source[columns])
        scaled = model[:-1].transform(source[columns].iloc[sample_rows])
        scores.append(silhouette_score(scaled, labels[sample_rows]))
    set_results.append({
        "Feature set": name,
        "Number of features": len(columns),
        "Silhouette (real data)": scores[0],
        "Silhouette (shuffled copy)": scores[1],
        "Gap": scores[0] - scores[1],
    })

display(pd.DataFrame(set_results).round(4))
```

## 4.2 Preprocessing
K-Means groups points by **distance**, so the three features are scaled to **mean 0** and **standard deviation 1** using the same imputer and scaler as in 2.2.3.

```python
final_features = clustering_feature_sets["Traffic load and road work"]

prep_cluster = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

X_scaled = prep_cluster.fit_transform(clustering_data[final_features])

X_reference = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
]).fit_transform(shuffled_data[final_features])

display(pd.DataFrame(X_scaled, columns=final_features).describe().loc[["mean", "std"]].round(3))
```

## 4.3 Test Different Numbers of Clusters
K-Means was fitted for K = 2 to 10 on the scaled data and on the shuffled copy. For each K the table shows inertia (the total squared distance from points to their centroid), the silhouette score on a sample of 5,000 incidents, and the Davies-Bouldin index (lower is better).

```python
from sklearn.metrics import davies_bouldin_score

k_values = list(range(2, 11))
k_rows = []

for i in range(len(k_values)):
    k = k_values[i]

    model = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = model.fit_predict(X_scaled)

    reference_model = KMeans(n_clusters=k, n_init=10, random_state=42)
    reference_labels = reference_model.fit_predict(X_reference)

    k_rows.append({
        "K": k,
        "Inertia": model.inertia_,
        "Shuffled inertia": reference_model.inertia_,
        "Silhouette": silhouette_score(X_scaled[sample_rows], labels[sample_rows]),
        "Shuffled silhouette": silhouette_score(X_reference[sample_rows], reference_labels[sample_rows]),
        "Davies-Bouldin": davies_bouldin_score(X_scaled, labels),
    })

k_results = pd.DataFrame(k_rows)
display(k_results.round(4))
```

##4.4 Elbow Method
Inertia always falls as K increases, so the elbow is the point where adding clusters stops helping much. To measure: both axes are scaled from 0 to 1 and the elbow is the K farthest from the straight line joining the first and last points.

```python
x_values = k_results["K"].to_numpy(dtype=float)
y_values = k_results["Inertia"].to_numpy(dtype=float)

x_norm = (x_values - x_values.min()) / (x_values.max() - x_values.min())
y_norm = (y_values - y_values.min()) / (y_values.max() - y_values.min())

elbow_distance = np.abs(x_norm + y_norm - 1) / np.sqrt(2)
k_elbow = int(x_values[np.argmax(elbow_distance)])

plt.figure(figsize=(8, 5))
plt.plot(k_results["K"], k_results["Inertia"], marker="o", label="Real data")
plt.plot(k_results["K"], k_results["Shuffled inertia"], marker="o", linestyle="--", label="Shuffled copy")
plt.axvline(k_elbow, color="red", linestyle=":", label=f"Elbow at K = {k_elbow}")
plt.title("Elbow Method: Inertia vs K")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.legend()
plt.tight_layout()
plt.show()

print("Elbow at K =", k_elbow)
print("Inertia drop from K-1 to K (%):", dict(zip(k_values[1:], (-np.diff(y_values) / y_values[:-1] * 100).round(1).tolist())))
```

## 4.5 Silhouette Score
The silhouette score compares how close each incident is to its own cluster with how close it is to the nearest other cluster. Values near 1 mean well separated clusters and values near 0 mean overlapping clusters. The Davies-Bouldin index is shown next to it as a second check.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(k_results["K"], k_results["Silhouette"], marker="o", label="Real data")
axes[0].plot(k_results["K"], k_results["Shuffled silhouette"], marker="o", linestyle="--", label="Shuffled copy")
axes[0].set_title("Average Silhouette Score")
axes[0].legend()

axes[1].plot(k_results["K"], k_results["Davies-Bouldin"], marker="o", color="tab:green")
axes[1].set_title("Davies-Bouldin Index")

for i in range(len(axes)):
    axes[i].set_xlabel("Number of clusters (K)")

plt.tight_layout()
plt.show()

k_silhouette = int(k_results.loc[k_results["Silhouette"].idxmax(), "K"])
k_db = int(k_results.loc[k_results["Davies-Bouldin"].idxmin(), "K"])
print("Best K by silhouette:", k_silhouette)
print("Best K by Davies-Bouldin:", k_db)
```

## 4.6 Choosing the number of clusters
The rule was fixed before looking for K: take the smallest K whose silhouette score is within 0.01 of the best, then check it against the elbow (a small K is preferred because fewer groups are easier to understand and act on).

```python
threshold = k_results["Silhouette"].max() - 0.01

optimal_k = None
for i in range(len(k_results)):
    if optimal_k is None and k_results.loc[i, "Silhouette"] >= threshold:
        optimal_k = int(k_results.loc[i, "K"])

print("Elbow K:", k_elbow)
print("Best silhouette K:", k_silhouette)
print("Smallest K within 0.01 of best silhouette:", optimal_k)
print("Lowest Davies-Bouldin K:", k_db)

chosen = k_results[k_results["K"] == optimal_k].iloc[0]
print(f"Selected K = {optimal_k}: silhouette {chosen['Silhouette']:.3f} (shuffled copy {chosen['Shuffled silhouette']:.3f}), Davies-Bouldin {chosen['Davies-Bouldin']:.3f}")
```

## 4.7 Final Clustering Model
The model was refitted with the chosen K and 20 random starts. K-Means numbers clusters in no particular order, so the clusters are renumbered from 1 to K by a risk index, which is the average of the three scaled centroid values. The index uses only the clustering features. The centroids are converted back to real units so they can be read.

```python
final_model = KMeans(n_clusters=optimal_k, n_init=20, random_state=42)
raw_labels = final_model.fit_predict(X_scaled)

order = np.argsort(final_model.cluster_centers_.mean(axis=1))
mapping = np.zeros(optimal_k, dtype=int)
for new_id in range(optimal_k):
    mapping[order[new_id]] = new_id + 1

df_fs["Cluster"] = mapping[raw_labels]

centroids_scaled = final_model.cluster_centers_[order]
centroids_real = prep_cluster.named_steps["scaler"].inverse_transform(centroids_scaled)

centroid_table = pd.DataFrame(centroids_real, columns=final_features, index=range(1, optimal_k + 1))
centroid_table["Incidents"] = df_fs["Cluster"].value_counts().sort_index()
centroid_table["Share (%)"] = centroid_table["Incidents"] / len(df_fs) * 100
centroid_table.index.name = "Cluster"
display(centroid_table.round(1))
```

## 4.8 Cluster Profiles
This compares the clusters on the three clustering features and on variables that were not used for clustering. Each cell shows how many standard deviations the cluster mean is above or below the overall mean.

```python
profile_columns = final_features + [
    "Congestion_Index", "Average_Speed", "Visibility", "Precipitation",
    "Wind_Speed", "Road_Work", "Peak_Hour"
]

profile = df_fs.groupby("Cluster")[profile_columns].mean()
profile_z = (profile - df_fs[profile_columns].mean()) / df_fs[profile_columns].std()

plt.figure(figsize=(12, 4.5))
sns.heatmap(profile_z, annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1.5, vmax=1.5)
plt.title("Cluster Means in Standard Deviations from the Overall Mean")
plt.tight_layout()
plt.show()
```

## 4.10 Comparison With Severity
Severity was not used to build the clusters. It is used here only to check them. The tables show the severity mix in each cluster and then the share of Severity 4 incidents in clear and adverse weather (Fog, Rain, Snow, Storm) within each cluster.

```python
from sklearn.metrics import normalized_mutual_info_score

severity_mix = pd.crosstab(df_fs["Cluster"], df_fs["Severity"], normalize="index") * 100
severity_mix.columns = ["Severity 1 (%)", "Severity 2 (%)", "Severity 3 (%)", "Severity 4 (%)"]
severity_mix["Mean severity"] = df_fs.groupby("Cluster")["Severity"].mean()
display(severity_mix.round(1))

print("NMI between clusters and Severity:", round(normalized_mutual_info_score(df_fs["Severity"], df_fs["Cluster"]), 3))

adverse = df_fs["Weather_Condition"].isin(["Fog", "Rain", "Snow", "Storm"])
is_severe = (df_fs["Severity"] == 4)

weather_effect = pd.DataFrame({
    "Clear, cloudy or windy: Severity 4 (%)": is_severe[~adverse].groupby(df_fs.loc[~adverse, "Cluster"]).mean() * 100,
    "Adverse weather: Severity 4 (%)": is_severe[adverse].groupby(df_fs.loc[adverse, "Cluster"]).mean() * 100,
})
display(weather_effect.round(1))
```

4.11 Hotspot Analysis
A hotspot is a place with more high risk incidents than chance would explain. High risk here means the two highest ranked clusters. State, City and District shares are compared with the overall share using z-scores and a Bonferroni limit.

```python
from scipy.stats import norm

df_fs["High_risk"] = df_fs["Cluster"] >= optimal_k - 1
overall_rate = df_fs["High_risk"].mean()
print(f"High-risk clusters: {optimal_k - 1} and {optimal_k} ({overall_rate * 100:.1f}% of incidents)")

group_columns = ["State", "City", "District"]
hotspot_rows = []

for column in group_columns:
    grouped = df_fs.groupby(column)["High_risk"].agg(["sum", "count"])
    shares = grouped["sum"] / grouped["count"]
    z_scores = (shares - overall_rate) / np.sqrt(overall_rate * (1 - overall_rate) / grouped["count"])
    limit = norm.ppf(1 - 0.05 / (2 * len(grouped)))
    hotspot_rows.append({
        "Grouping": column,
        "Groups tested": len(grouped),
        "Lowest share (%)": shares.min() * 100,
        "Highest share (%)": shares.max() * 100,
        "Largest |z|": z_scores.abs().max(),
        "Bonferroni limit": limit,
        "Hotspots flagged": int((z_scores.abs() > limit).sum()),
    })

display(pd.DataFrame(hotspot_rows).round(2))
```

## 4.12 Practical Use of the Clusters
Each cluster is described from its centroid (high if at least 0.5 standard deviations above average, low if at least 0.5 below) and given a monitoring action by its rank. The table also shows how many Severity 4 incidents are captured if the highest-risk clusters are checked first. Lift is the share of Severity 4 incidents captured divided by the share of incidents checked, and 1.0 means no better than random. Finally, three invented situations are placed into clusters with the fitted scaler and model.

```python
feature_words = ["traffic density", "vehicle count", "road-work presence"]
tier_rows = []
cumulative_incidents = 0
cumulative_severe = 0
total_severe = (df_fs["Severity"] == 4).sum()

for rank in range(optimal_k, 0, -1):
    parts = []
    for j in range(len(final_features)):
        z = centroids_scaled[rank - 1, j]
        if z >= 0.5:
            parts.append("high " + feature_words[j])
        elif z <= -0.5:
            parts.append("low " + feature_words[j])

    if len(parts) == 0:
        description = "mid-range on all three"
    else:
        description = ", ".join(parts)

    if rank >= optimal_k - 1:
        action = "High: priority monitoring, pre-position response, active signal and speed control"
    elif rank <= 2:
        action = "Low: routine monitoring"
    else:
        action = "Medium: enhanced monitoring in adverse weather and peak hours"

    members = df_fs[df_fs["Cluster"] == rank]
    cumulative_incidents = cumulative_incidents + len(members)
    cumulative_severe = cumulative_severe + (members["Severity"] == 4).sum()

    tier_rows.append({
        "Cluster": rank,
        "Profile": description,
        "Action": action,
        "Incidents checked (%)": cumulative_incidents / len(df_fs) * 100,
        "Severity 4 captured (%)": cumulative_severe / total_severe * 100,
        "Lift": (cumulative_severe / total_severe) / (cumulative_incidents / len(df_fs)),
    })

pd.set_option("display.max_colwidth", 80)
display(pd.DataFrame(tier_rows).set_index("Cluster").round(2))

new_situations = pd.DataFrame({
    "Situation": ["Quiet rural road", "Heavy flow on black-spot road", "Light traffic near road works"],
    "Traffic_Density": [20, 95, 25],
    "Vehicle_Count": [4, 24, 6],
    "Road_Work": [1, 18, 17],
})

new_raw = final_model.predict(prep_cluster.transform(new_situations[final_features]))
new_situations["Assigned cluster"] = mapping[new_raw]
display(new_situations)
```
